```python
import matplotlib.pyplot as plt
from torchvision import io, utils
import lightly_train

IMAGE_PATH = "../../SVRDD_YOLO/images/train/09002200121901211022275042K_1296262200_483053600.jpg"

model = lightly_train.load_model("/home/andre/TCC/models/DINOv3/output_dinov3/exported_models/exported_best.pt")
labels, boxes, scores = model.predict(IMAGE_PATH).values()

# Visualize predictions.
image_with_boxes = utils.draw_bounding_boxes(
    image=io.read_image(IMAGE_PATH),
    boxes=boxes,
    labels=[model.classes[i.item()] for i in labels],
)

fig, ax = plt.subplots(figsize=(30, 30))
ax.imshow(image_with_boxes.permute(1, 2, 0))
fig.savefig("predictions.png")
```

# Training

```python
import torch
import lightly_train

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

lightly_train.train_object_detection(
    out="output_dinov3",
    model="dinov3/convnext-tiny-ltdetr-coco",
    data={
        "path": "/workspace/SVRDD_YOLO",
        "train": "images/train",
        "val": "images/val",
        "names": {
            0: "longitudinal crack",
            1: "transverse crack",
            2: "alligator crack",
            3: "pothole",
            4: "manhole cover",
            5: "longitudinal patch",
            6: "transverse patch",
        }
    },
    overwrite=True,
    #device=device,
    batch_size=8,
    steps=50000
)
```
