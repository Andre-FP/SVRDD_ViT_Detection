```python
from ultralytics import YOLO

# Load a model
model = YOLO('yolo11m.pt')  # load a pretrained model (recommended for training)
```

```python
results = model.train(
    data="/workspace/SVRDD_YOLO/data.yaml",
    epochs=100,
    imgsz=1024,
    batch=8,
    name="SVRDD_YOLO11m",
    cache=True,
    half=False,
    save_period=1,
    amp=True,
)
```
