```python
from ultralytics import YOLO

# Load a model
model = YOLO('yolov5s6u.pt')  # load a pretrained model (recommended for training)
```

```python
results = model.train(
    data="./SVRDD_YOLO/data.yaml",
    epochs=100,
    imgsz=1024,
    batch=16,
    name="YOLOv5_output",
    cache=True,
    half=True,
    save_period=1,
    amp=True,
)
```
